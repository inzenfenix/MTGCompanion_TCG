```python
# requires_grad=True le dice a PyTorch que queremos calcular gradientes
x = torch.tensor(2.0, requires_grad=True)
w = torch.tensor(3.0, requires_grad=True)
b = torch.tensor(1.0, requires_grad=True)
# Función: y = w*x + b (una neurona sin activación)
y = w * x + b
print("y =", y.item()) # 7.0
# Calcular gradientes respecto a y
y.backward()
# dy/dx = w = 3.0
# dy/dw = x = 2.0
# dy/db = 1.0
print("grad x:", x.grad) # tensor(3.)
print("grad w:", w.grad) # tensor(2.)
print("grad b:", b.grad) # tensor(1.)
# IMPORTANTE: limpiar gradientes antes del siguiente paso
x.grad.zero_()
w.grad.zero_()
b.grad.zero_()
```

```python
import torch
import torch.nn as nn
# Una capa lineal: 3 entradas → 1 salida
neurona = nn.Linear(in_features=3, out_features=1)
# Ver parámetros inicializados aleatoriamente
print("Pesos:", neurona.weight) # shape: (1, 3)
print("Bias:", neurona.bias) # shape: (1,)
# Forward pass con un batch de 5 ejemplos
x = torch.rand(5, 3) # 5 muestras, 3 features
y_pred = neurona(x)
print("Salida:", y_pred.shape) # torch.Size([5, 1])
print(y_pred)
```

```python
import torch

# 1. Definimos los puntos x donde queremos evaluar la derivada
# Es CRUCIAL pasar `requires_grad=True` para que PyTorch rastree las operaciones en el grafo
valores_x = [1.0, 2.0, 3.0]

print(f"{'x':<5} | {'Derivada Autograd':<20} | {'Derivada Analítica':<20} | {'¿Coinciden?':<10}")
print("-" * 65)

for val in valores_x:
    # Creamos el tensor para el punto x actual
    x = torch.tensor(val, requires_grad=True)
    
    # 2. Construimos el grafo computacional para f(x) = 2x^2 - 3x + 1
    # PyTorch registra internamente cada operación (potencia, multiplicación, resta, suma)
    y = 2 * (x ** 2) - 3 * x + 1
    
    # 3. Backward pass: Calculamos el gradiente (df/dx)
    y.backward()
    
    # El resultado se almacena en el atributo .grad del tensor x
    derivada_autograd = x.grad.item()
    
    # 4. Cálculo manual usando la derivada analítica f'(x) = 4x - 3
    derivada_analitica = 4 * val - 3
    
    # Verificación
    coinciden = "Sí" if derivada_autograd == derivada_analitica else "No"
    
    print(f"{val:<5} | {derivada_autograd:<20} | {derivada_analitica:<20} | {coinciden:<10}")
```
